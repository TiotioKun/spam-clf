```python
import subprocess, sys, time, os

subprocess.run(["git", "clone", "-q",
                "https://github.com/TiotioKun/spam-clf.git",
                "/content/spam-clf"], check=False)
os.chdir("/content/spam-clf")

subprocess.run([sys.executable, "-m", "pip", "install", "-q",
                "fastapi", "uvicorn", "python-multipart", "jinja2",
                "lightgbm", "joblib"], check=False)

server = subprocess.Popen(
    [sys.executable, "-m", "uvicorn", "src.app.main:app", "--port", "8000"],
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)

time.sleep(15)

from google.colab.output import eval_js
url = eval_js("google.colab.kernel.proxyPort(8000)")
print("Интерфейс:   ", url)
print("Документация:", url + "docs")
```
